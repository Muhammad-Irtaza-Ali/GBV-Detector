```python
import json
import numpy as np

from datasets import Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    TrainingArguments,
    Trainer,
    DataCollatorWithPadding
)
```

```python
#load json
json_file = "/content/drive/MyDrive/GBV Project/GBV_detect_Urdu.json"

with open(json_file,"r",encoding="utf-8") as f:
    data = json.load(f)
```

```python
#Create binary labels       Use the is_gbv field.
texts = []
labels = []

for sample in data:

    if sample["text"].strip() == "":
        continue

    texts.append(sample["text"])

    labels.append(int(sample["is_gbv"]))
```

```python
#Dont Run this
import json
from collections import defaultdict

# Load JSON
json_file = "/content/drive/MyDrive/GBV Project/GBV_detect_Urdu.json"

with open(json_file, "r", encoding="utf-8") as f:
    data = json.load(f)

invalid_ids = defaultdict(list)

for sample in data:
    value = sample.get("is_gbv")

    if not isinstance(value, bool):
        invalid_ids[value].append(sample.get("id"))

print("Invalid values and corresponding article IDs:\n")

for value, ids in invalid_ids.items():
    print(f"{value}:")
    print(ids)
    print(f"Total: {len(ids)}\n")
```

```python
# after binarization of labels start from there
#Train/Test Split
train_texts, test_texts, train_labels, test_labels = train_test_split(
    texts,
    labels,
    test_size=0.15,
    random_state=42,
    stratify=labels
)
```

```python
#Convert to HuggingFace Dataset
train_dataset = Dataset.from_dict({
    "text": train_texts,
    "label": train_labels
})

test_dataset = Dataset.from_dict({
    "text": test_texts,
    "label": test_labels
})
```

```python
#load tokenizer
model_name = "xlm-roberta-base"
```

```python
#tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
#Tokenization
def tokenize(batch):

    return tokenizer(
        batch["text"],
        truncation=True,
        padding=False,
        max_length=256
    )

train_dataset = train_dataset.map(tokenize, batched=True)
test_dataset = test_dataset.map(tokenize, batched=True)
```

```python
#Load Model
model = AutoModelForSequenceClassification.from_pretrained(
    model_name,
    num_labels=2
)
```

```python
def compute_metrics(eval_pred):

    logits, labels = eval_pred

    predictions = np.argmax(logits, axis=1)

    precision, recall, f1, _ = precision_recall_fscore_support(
        labels,
        predictions,
        average="binary"
    )

    acc = accuracy_score(labels, predictions)

    return {
        "accuracy": acc,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }
```

```python
training_args = TrainingArguments(

    output_dir="./gbv_model",

    learning_rate=2e-5,

    per_device_train_batch_size=8,

    per_device_eval_batch_size=8,

    num_train_epochs=5,

    weight_decay=0.01,

    eval_strategy="epoch",

    save_strategy="epoch",

    load_best_model_at_end=True,

    metric_for_best_model="f1",

    greater_is_better=True,

    logging_steps=20,

    fp16=True
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    processing_class=tokenizer,
    data_collator=DataCollatorWithPadding(tokenizer=tokenizer),
    compute_metrics=compute_metrics,
)
```

```python
#finetune
trainer.train()
```

```python
trainer.evaluate()
```

```python
save_path="/content/drive/MyDrive/GBV Project/GBV_BERT"

trainer.save_model(save_path)

tokenizer.save_pretrained(save_path)

print("Model saved.")
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load model
model.to(device)
model.eval()

text = "شوہر نے بیوی کو شدید تشدد کا نشانہ بنایا۔"

inputs = tokenizer(
    text,
    return_tensors="pt",
    truncation=True,
    max_length=256
)

# Move all inputs to GPU
inputs = {k: v.to(device) for k, v in inputs.items()}

with torch.no_grad():
    outputs = model(**inputs)

prediction = outputs.logits.argmax(dim=1).item()

print("Prediction:", prediction)
print("GBV" if prediction == 1 else "Not GBV")
```
